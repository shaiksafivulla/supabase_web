```python
import pandas as pd

hotels = []

for title_tag in soup.find_all(
    "h3", class_="listingHotelDescription__hotelName"
):

    card = title_tag.find_parent(
        class_="hotelCardListing"
    )

    if card is None:
        card = title_tag.parent.parent.parent


    title = title_tag.get_text(" ", strip=True)


    address_tag = card.find(
        class_="listingHotelDescription__hotelAddress"
    )
    address = address_tag.get_text(" ", strip=True) if address_tag else None


    price_tag = card.find(
        class_="listingPrice__finalPrice"
    )
    price = price_tag.get_text(" ", strip=True) if price_tag else None

    count_tag = card.find(
        class_="hotelRating__ratingSummary"
    )
    rating_count = count_tag.get_text(" ", strip=True) if count_tag else None

    hotels.append({
        "Title": title,
        "Address": address,
        "Price": price,
        "Rating Count": rating_count
    })

df = pd.DataFrame(hotels)

display(df)
```
